```python
data_with_missing = {
    'cars': ["BMW", "Volvo", "Ford", "Tesla", None],
    'passings': [3, 7, 2, None, 4]
}

df_data = pd.DataFrame(data_with_missing)

print(df_data)
```

```python
df_data.isnull()
```

```python
df_data.isnull().sum()
```

```python
# dropna - remove rows with missing values
df_drop = df_data.dropna()
print(df_drop)
```

```python
# fillna - fill missing values
df_fill = df_data.fillna({'cars': 'Unknown', 'passings': df_data['passings'].mean()})
print(df_fill)
```

```python
# add a new column
df['price'] = [50000, 40000, 30000]
print(df)
```

```python
# drop a column
df_no_price = df.drop(columns=['price'])
print(df_no_price)
```

```python
# rename columns
df_renamed = df.rename(columns={'cars': 'car_name'})
print(df_renamed)
```

```python
# sort_values
print(df.sort_values('passings', ascending=False))
```

```python
# iloc - select by position
print(df.iloc[0])
print(df.iloc[0:2])
```

```python
# select columns
print(df[['cars', 'price']])
```

```python
# value_counts
print(df['cars'].value_counts())
```

```python
# groupby
sales = pd.DataFrame({
    'brand': ["BMW", "BMW", "Ford", "Ford", "Volvo"],
    'sold': [5, 3, 8, 2, 4]
})

print(sales.groupby('brand')['sold'].sum())
```

```python
# apply
df['price_k'] = df['price'].apply(lambda x: x / 1000)
print(df)
```

```python
# merge
owners = pd.DataFrame({
    'cars': ["BMW", "Volvo", "Ford"],
    'owner': ["Ali", "Sara", "John"]
})

print(pd.merge(df, owners, on='cars'))
```

```python
# save to CSV and read it back
df.to_csv('cars.csv', index=False)

df_csv = pd.read_csv('cars.csv')
print(df_csv)
```
