```python
# missing values and duplicates
print(df.isnull().sum())
print("duplicates:", df.duplicated().sum())
```

```python
# grade distribution
print(df['grade'].value_counts().sort_index())
```

```python
# average stats per grade
df.groupby('grade')[['weekly_self_study_hours', 'attendance_percentage', 'class_participation', 'total_score']].mean()
```

```python
# top 10 students by score
df.sort_values('total_score', ascending=False).head(10)
```

```python
# students with low score
low = df[df['total_score'] < 50]
print("students below 50:", len(low))
```

```python
# study hours groups
df['study_group'] = pd.cut(df['weekly_self_study_hours'], bins=[0, 10, 20, 30, 40], include_lowest=True, labels=['0-10', '10-20', '20-30', '30-40'])
df.groupby('study_group', observed=True)['total_score'].mean()
```

```python
# attendance groups
df['attendance_group'] = pd.cut(df['attendance_percentage'], bins=[50, 70, 85, 100], include_lowest=True, labels=['low', 'medium', 'high'])
df.groupby('attendance_group', observed=True)['total_score'].mean()
```

```python
# which feature is most correlated with total_score
df_columns.corr()['total_score'].drop('total_score').sort_values(ascending=False)
```

```python
# percentage of students with perfect score
perfect = (df['total_score'] == 100).mean() * 100
print(f"perfect score: {perfect:.2f}%")
```

```python
# pass / fail column
df['passed'] = np.where(df['total_score'] >= 60, 'pass', 'fail')
print(df['passed'].value_counts(normalize=True) * 100)
```
