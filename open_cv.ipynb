```python
import cv2 as cv
```

```python
print(cv.__version__)
```

```python
from google.colab.patches import cv2_imshow

img = cv.imread("/content/dog-image1.jpeg")
cv2_imshow(img)
```

```python
# image shape: (height, width, channels)
print("shape:", img.shape)
print("size:", img.size)
print("dtype:", img.dtype)
```

```python
# grayscale
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
# resize
resized = cv.resize(img, (300, 200))
cv2_imshow(resized)
```

```python
# crop (rows, columns)
cropped = img[50:250, 100:350]
cv2_imshow(cropped)
```

```python
# rotate 90 degrees
rotated = cv.rotate(img, cv.ROTATE_90_CLOCKWISE)
cv2_imshow(rotated)
```

```python
# flip horizontally
flipped = cv.flip(img, 1)
cv2_imshow(flipped)
```

```python
# blur
blurred = cv.GaussianBlur(img, (15, 15), 0)
cv2_imshow(blurred)
```

```python
# edge detection
edges = cv.Canny(gray, 100, 200)
cv2_imshow(edges)
```

```python
# threshold (black and white)
_, thresh = cv.threshold(gray, 127, 255, cv.THRESH_BINARY)
cv2_imshow(thresh)
```

```python
# draw shapes and text
drawing = img.copy()
cv.rectangle(drawing, (50, 50), (250, 250), (0, 255, 0), 3)
cv.circle(drawing, (300, 150), 50, (255, 0, 0), 3)
cv.line(drawing, (0, 0), (img.shape[1], img.shape[0]), (0, 0, 255), 2)
cv.putText(drawing, "Dog", (50, 40), cv.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
cv2_imshow(drawing)
```

```python
# split color channels (OpenCV uses BGR order)
b, g, r = cv.split(img)
cv2_imshow(b)
cv2_imshow(g)
cv2_imshow(r)
```

```python
# save the result
cv.imwrite("/content/dog-gray.jpeg", gray)
```

OPENCV video streaming codes

```python
print("hello world")

import cv2 as cv

cap = cv.VideoCapture(0, cv.CAP_AVFOUNDATION)

face_cascade = cv.CascadeClassifier(cv.data.haarcascades + "haarcascade_frontalface_default.xml")

while True:
    success, frame = cap.read() # True / False

    if not success: break

    faces = face_cascade.detectMultiScale(frame, 1.5, 4)
    # frame_gray = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)


    for (x, y, w, h) in faces:
        cv.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    cv.imshow("Real-time Face Detection", frame)

    # cv.imshow('Original', frame_gray)

    if(cv.waitKey(1) & 0xFF == ord('q')): break

cap.release()
cv.destroyAllWindows()
```
