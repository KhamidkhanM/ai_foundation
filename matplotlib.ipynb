```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 25, 30])

plt.plot(x, y)
plt.title("Basic Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([8, 20, 29, 15])

plt.plot(y, linestyle="dotted", color='green', marker='o')
plt.title("Basic Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
x = np.array([1, 2, 3, 4])
y1 = np.array([8, 20, 29, 15])
y2 = np.array([10, 20, 25, 5])

plt.plot(x, y1, linestyle="dotted", color='red', marker='s')
plt.plot(x, y2, linestyle="dotted", color='green', marker='d')
plt.title("Basic Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend(["Line 1", "Line 2"])
plt.show()
```

# Barchart with Numpy and Matplotlib

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([5, 9, 3, 7])

plt.bar(x, y, color='orange')
plt.title("Basic Barchart Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

# Scatter Plot with Numpy and Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([5, 4, 6, 3, 7])

plt.scatter(x, y, color='purple')
plt.title('Scatter Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

# Histogram with Numpy and MatplotLib

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([3, 4, 5, 6, 8, 8, 8, 9, 10])

plt.hist(data, bins=5, color='skyblue', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel('Values')
plt.ylabel('Count')
plt.show()
```

# Subplots (Multiple graphs together)

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y1 = np.array([2, 4, 6, 8])
y2 = np.array([3, 6, 9, 12])
y3 = np.array([3, 5, 9, 7])

plt.figure(figsize=(8, 4))

plt.subplot(1, 3, 1)
plt.plot(x, y1, color='red')
plt.title('Plot 1')

plt.subplot(1, 3, 2)
plt.bar(x, y2, color='blue')
plt.title('Plot 2')

plt.subplot(1, 3, 3)
plt.bar(x, y3, color='purple')
plt.title('Plot 3')

plt.tight_layout()
plt.show()
```
